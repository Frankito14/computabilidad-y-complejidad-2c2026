## Tarea 7: Máquinas Secuenciales

*1. Detector de la secuencia 01*
Diseñar una máquina que emita:
- Salida 1 cuando detecte la secuencia 01
- Salida 0 en cualquier otro caso

*Maquina secuencial*: Mealy

### Definición Formal

M = < ΣE, ΣS, Q, f, g >

- _ΣE: { 1, 0 }_ (alfabeto de símbolos de entrada)
- _ΣS: { 1, 0}_ (alfabeto de símbolos de salida)
- _Q: { q0, q1 }_ (conjunto finito no vacío de estados)
- _f: Q x ΣE -> Q_ (función de transición)
- _g: Q x ΣE -> ΣS_ (función de salida)

### Tabla de Transiciones

| f / g | 0 | 1 |
| ------- | -------- | -------- |
| >q0 | q1 / 0 | q0 / 0 |
| q1  | q1 / 0 | q0 / 1 |
