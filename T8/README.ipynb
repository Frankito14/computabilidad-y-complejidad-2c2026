## Tarea 8: Complejidad

Determinar complejidad espacial y temporal

*1. L = { wcw^-1 / w ∈ ∑* }*

- Entrada (n) = 2|w| + 1 
- |w| ≈ n / 2

### Algoritmo (Estrategia de Ping-Pong)

- Repetir (n / 2):
1. Marcar el primer simbolo sin procesar de la izquierda de C (w)
2. Ir hasta el ultimo simbolo sin procesar de la derecha de C (w^-1)
3. Si coincide, marcar el ultimo simbolo sin procesar de la derecha de C (w^-1)
4. Ir hasta el primer simbolo sin procesar de la izquierda de C (w)

### Complejidad Temporal (Cantidad Movimientos)

- Cantidad de viajes = (n / 2)
- Largo de la cinta = n
- Total de movimientos = (n / 2) * n = n^2 / 2
- Complejidad: *T(n) = O(n^2) (Cuadratica)*

### Complejidad Espacial (Cantidad Celdas)

- Primer celda posible: Primer simbolo sin procesar de la izquierda de C (w)
- Ultima celda posible: Blanco despues del ultimo simbolo ya procesado de la derecha de C (w^-1)
- Cantidad celdas usadas = (n + 1)
- Complejidad: *S(n) = O(n) (Lineal)*


*2. f((ab)^n) = (ab)^n+1*

- Longitud de la entrada = 2n
- Ejemplo: abab (n = 2) -> N = 4 

### Algoritmo (Estrategia de Push)

1. Ir hasta el final de la cadena (2n movimientos)
2. Mover a la derecha y escribir un simbolo a (1 movimiento)
2. Mover a la derecha y escribir un simbolo b (1 movimiento)


### Complejidad Temporal (Cantidad Movimientos)

- Cantidad de movimientos para llegar al final: 2n
- Cantidad de movimientos para escribir "ab" al final de la cadena: 2
- Total de movimientos = 2n + 2
- Complejidad: *T(n) = O(n) (Lineal)*

### Complejidad Espacial (Cantidad Celdas)

- Recorremos toda la entrada: 2n celdas
- Recorremos 2 celdas mas para escribir los simbolos al final de la entrada: 2 celdas
- Total de celdas = 2n + 2
- Complejidad: *S(n) = O(n) (Lineal)*
